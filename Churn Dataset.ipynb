```python
df = pd.read_csv(r"Churn_Dataset2.csv")
df
```

```python
df.isnull().sum()
```

```python
for x in ["total eve calls", "total night calls", "total intl calls"]:
    df[x] = df[x].fillna(df[x].median())
```

```python
df["total day calls"] = df["total day calls"].fillna(df["total intl calls"].mean())
```

```python
df.isnull().sum()
```

```python
df1 = df.select_dtypes(include = ["object"])
```

```python
df1.columns
```

```python
cols = ["account length","area code","number vmail messages",
       "total day minutes","total day calls", "total day charge",
       "total eve minutes", "total eve calls","total eve charge",
        "total night minutes", "total night calls","total night charge",
       "total intl minutes", "total intl calls", "total intl charge", "customer service calls"]

plt.figure(figsize = (10,8))
plt.boxplot(df[cols])
plt.xticks(ticks = range(1,len(cols)+1), labels= cols, rotation=90)
plt.show()
```

```python
Q1 = df[cols].quantile(0.25)
Q3 = df[cols].quantile(0.75)
IQR = Q3 - Q1
df_cleaned = df[
    ~(
        ((df[cols] < (Q1 - 1.5 * IQR)) | 
        (df[cols] > (Q3 + 1.5 * IQR)))
        .any(axis=1))]
print("Before:", df.shape)
print("After:", df_cleaned.shape)
```
